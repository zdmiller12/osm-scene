# NumPy STL

> https://numpy-stl.readthedocs.io/en/latest/usage.html

```python
import pickle
from pathlib import Path

import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
from stl import mesh

FIGURE_PICKLE = Path("stl/Cube.figure.pickle").resolve()
```

## Load Pre-defined STL, Create Figure, and Save

```python
figure = plt.figure()
axes = figure.add_subplot(111, projection="3d")

# load an example STL file and add the vectors
your_mesh = mesh.Mesh.from_file("stl/Cube.stl")
_ = axes.add_collection3d(mplot3d.art3d.Poly3DCollection(your_mesh.vectors))

# auto-scale to the mesh size
scale = your_mesh.points.flatten()
_ = axes.auto_scale_xyz(scale, scale, scale)

# save figure (just to show example of saving interactive plot by pickling)
with FIGURE_PICKLE.open("wb") as p:
    pickle.dump(figure, p)
display(f"Saved {FIGURE_PICKLE}")
plt.close()
```

## Load and Show Figure

```python
%matplotlib ipympl

with FIGURE_PICKLE.open("rb") as p:
    figure_loaded = pickle.load(p)
figure_loaded.show()
```
